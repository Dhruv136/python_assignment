Section B aka pandas

```python
import pandas as pd
#<<<<<<<<<<<<<<< Question 5 >>>>>>>>>>>>>>

data = {
    "Name": ["dhruv", "gojo", "satoru", "kenji", "kamikaze",
             "vegeta", "siyu", "giyu", "tanjiro", "muzan"],

    "Age": [20, 200, 332, 43, 43, 45, 64, 67, 89, 100],

    "Marks": [309, 45, 99, 345, 56, 343, 566, 99, 4332, 4322],

    "City": ["Ramgarh", "sihuah", "goregaon", "kaalegaon", "hehegaon",
             "Delhi", "hyderbad", "nasik", "chirawa", "surajgarh"]
}

df = pd.DataFrame(data)

print("First 5 rows:")
print(df.head())

print("Statistics:")
print(df.describe())

print("Students with Marks > 75:")
print(df[df["Marks"] > 75])

print("Sorted by Marks (Descending):")
print(df.sort_values(by="Marks", ascending=False))
```

```python
#<<<<<<<<<<<<<<<<< Question 6 >>>>>>>>>>>>>>>>>>>>>>
data = {
    "Name": ["Dhruv", "Ishika", "Gojo", "Kenji", "Vegeta"],
    "Marks": [85, np.nan, 92, np.nan, 78],
    "City": ["Pilani", "Delhi", np.nan, "Mumbai", np.nan]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("Missing Values:")
print(df.isnull())

mean_marks = df["Marks"].mean()
df["Marks"] = df["Marks"].fillna(mean_marks)

print("After Filling Missing Marks with Mean:")
print(df)

df_city= df.dropna(subset=["City"])

print("After Dropping Rows with Missing City:")
print(df_city)
```

```python
# <<<<<<<<< Question 7 >>>>>>>>>>>>>>>>>

df = pd.read_csv("sales.csv")

category_sales = df.groupby("Category").agg(
    Total_Sales=("Sales", "sum")
)

print("Total Sales by Category:")
print(category_sales)

top3 = df.sort_values("Sales", ascending=False).head(3)

print("Top 3 Products by Sales:")
print(top3)
```

```python
#<<<<<<<<<<<<<<<< Question 8 >>>>>>>>>>>>>>>>>>>
df1 = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105],
    "Name": ["Dhruv", "Gojo", "Kenji", "Vegeta", "Tanjiro"],
    "Department": ["CSE", "IT", "ECE", "ME", "CSE"]
})
df2 = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105],
    "Marks": [95, 78, 88, 65, 92],
    "Grade": ["A", "B", "A", "C", "A"]
})

merged_df = pd.merge(df1, df2, on="StudentID")

print("Merged DataFrame:")
print(merged_df)

print("Students with Grade A:")
print(merged_df[merged_df["Grade"] == "A"])
```

Section C – Data Visualization 

```python
#<<<<<<<<<<<<<<<< Question 9 >>>>>>>>>>>>>>>>
import matplotlib.pyplot as plt

subjects = ["Math", "Science", "English", "Computer", "History"]
marks = [85, 90, 78, 95, 88]
plt.plot(subjects, marks, marker='o')
plt.title("Marks Trend Across 5 Subjects")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)
plt.show()
students = ["Dhruv", "Gojo", "Kenji", "Vegeta", "Tanjiro"]
student_marks = [85, 92, 78, 88, 95]
plt.bar(students, student_marks)
plt.title("Marks of 5 Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#<<<<<<<<<<<<<<<< Question 10 >>>>>>>>>>>>>>>>>>

data = {
    "Math": [85, 90, 78, 88, 95],
    "Science": [80, 92, 75, 85, 89],
    "English": [88, 79, 82, 91, 87],
    "Marks": [84, 87, 78, 88, 90],
    "Category": ["A", "A", "B", "A", "B"]
}

df = pd.DataFrame(data)

df["Marks"].plot(kind="hist", bins=5, title="Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

df["Marks"].plot(kind="box", title="Box Plot of Marks")
plt.ylabel("Marks")
plt.show()

df["Category"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Category-wise Percentage"
)
plt.ylabel("")
plt.show()
```

```python
#<<<<<<<<<<<<<<<<< Question 11 >>>>>>>>>>>>>>>
import seaborn as sns
np.random.seed(1)

df = pd.DataFrame({
    "Age": np.random.randint(22, 60, 50),
    "Salary": np.random.randint(25000, 100000, 50),
    "Department": np.random.choice(["HR", "IT", "Sales", "Finance"], 50),
    "Experience": np.random.randint(1, 20, 50)
})

print(df.head())

plt.figure(figsize=(6,4))
sns.scatterplot(data=df, x="Age", y="Salary", hue="Department")
plt.title("Age vs Salary by Department")
plt.show()

plt.figure(figsize=(6,4))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

plt.figure(figsize=(6,4))
sns.barplot(data=df, x="Department", y="Salary")
plt.title("Average Salary per Department")
plt.show()
```
